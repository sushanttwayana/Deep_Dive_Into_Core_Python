#### ENUM

An Enum is a way to group related, constant values together under one type — so your code becomes cleaner, safer, and more readable.

```python
# Without using Enum

# Without Enum
STATUS_PENDING = 1
STATUS_IN_PROGRESS = 2
STATUS_COMPLETED = 3

def update_status(status):
    if status == STATUS_PENDING:
        print("Task is pending")
    elif status == STATUS_COMPLETED:
        print("Task completed!")

update_status(3)  # Works, but not very clear
```

Problem?

You can pass any number (like 99) by mistake, and Python won’t complain.

You have to remember what each number means.

```python
# using enum

from enum import Enum

class Status(Enum):
    PENDING = 1
    IN_PROGRESS = 2
    COMPLETED = 3

def update_status(status):
    if status == Status.PENDING:
        print("Task is pending")
    elif status == Status.COMPLETED:
        print("Task completed!")

update_status(Status.COMPLETED)  # Much clearer!
```

#### Why this is better:

You can only use valid enum members (e.g., Status.PENDING)

Easy to read and maintain.

Gives structure — especially useful in large projects.

```python
# Accessing Enum members and values
print(Status.PENDING)          # Output: Status.PENDING
print(Status.PENDING.name)     # Output: PENDING
print(Status.PENDING.value)    # Output: 1
```

```python
# Iterating over Enums

for status in Status:
    print(status)
```

```python
# Comparing Enums

print(Status.PENDING == Status.PENDING)
print(Status.PENDING == Status.COMPLETED)
```

You can compare Enums only within the same class.

If you compare Enums from different classes, Python will raise an error — which is good because it avoids logical bugs.

```python
# Real life scenario exmaple in ordering systemm

from enum import Enum

class Order_Status(Enum):
    
    place = "placed"
    cook = "cooking"
    deliver = "delivered"
    cancel = "canceled"
    
def order_update(status: Order_Status):
    
    if status == Order_Status.place:
        print("Your Order is PLACED!!!")
    
    elif status == Order_Status.cook:
        print("Your Order is being prepared!!!")
        
    elif status == Order_Status.cancel:
        print("Your Order is canceled!!!")
    
    elif status == Order_Status.deliver:
        print("Your Order is delivered. Please, enjoy your meal")
    
    else:
        print("Please select proper order criteria")
        
order_update(Order_Status.cook)
```

**Enum best practices**

✅ Use Enums when:

- You have a fixed set of related constants (like states, modes, or categories).

- You want type safety (to prevent invalid values).

- You want clear, self-documenting code.

❌ Don’t use Enums for:

- Values that change frequently.

- Data that comes from a database or API dynamically.

### auto() feature

You can let Python automatically assign values:



```python
from enum import Enum, auto

class Priority(Enum):
    LOW = auto()
    MEDIUM = auto()
    HIGH = auto()

for p in Priority:
    print(p.name, p.value)
```
