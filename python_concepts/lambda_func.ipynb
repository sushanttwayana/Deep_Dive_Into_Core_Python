### What is a Python Lambda Function?
A lambda function is simply a small, anonymous, one-line function written in Python.

Think of it like a sticky note 📝 you use to write down a very short, simple mathematical rule, instead of writing a full letter (which would be a regular function, called with def).

* Anonymous: It doesn't have a formal name like functions created with the def keyword.

* One-Line: It can only contain a single expression (a calculation or operation) that it immediately returns.

* Syntax: It always starts with the keyword lambda.



```python
# Regular Function (The "Letter")	Lambda Function (The "Sticky Note")
def add_ten(x):	
    return x + 10

add_ten(10)

```

```python
add_ten = lambda x: x + 10

add_ten(20)
```

### Why Use Lambda Functions?
Lambda functions aren't meant to replace regular functions; they're used in specific situations where a tiny, temporary function is needed, especially as an argument to another function.

1. Using with sort() (The Most Common Use)
    * This is where lambda functions shine. Python's built-in functions like sorted() or list.sort() can take an optional argument called key. The key tells the function how to sort the items.

Scenario: You have a list of tuples (like coordinates or names with ages) and you want to sort them based on the second item in each tuple, not the first.

```python
# A list of people (Name, Age)
people = [('Alice', 30), ('Bob', 25), ('Charlie', 35)]

# We want to sort them by Age (the second item, index 1)

# Using a lambda function as the 'key'
sorted_by_age = sorted(people, key=lambda person: person[1])

print(sorted_by_age) 
# Output: [('Bob', 25), ('Alice', 30), ('Charlie', 35)]
```

The lambda function $\text{lambda person}: \text{person}[1]$ is a throwaway function that is used only by the sorted() function. For every person tuple it sees, it hands back only the age (person[1]) for sorting comparison, but the original tuple is what remains in the final sorted list.

```python
list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4]))
```

```python
list(map(lambda x: x * 2, [1, 2, 3]))
```

## Limitations (What it CANNOT do)
Since lambda is meant for simplicity, it has strict rules:

Only One Expression: You cannot have multiple lines of code, like an if/else block with two separate actions, or a for loop.

No Statements: You can't use complex Python statements like return, yield, pass, or assignment (=).

Readability: For anything complex, a regular def function is always better because it allows comments and multiple lines, making it easier for someone else (or your future self) to understand.

In short, use a lambda function when you need a quick, simple calculation right where you are using it, but always use a regular def function for any logic that needs more than one line.
